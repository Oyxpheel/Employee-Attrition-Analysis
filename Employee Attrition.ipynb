```python
#1. What does the company's workforce look like?
ABC_workforce= HR_employee_attrition[['Gender', 'Age' , 'Department']]
ABC_workforce['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Number of Employees')
plt.show()

ABC_workforce['Department'].value_counts().plot(kind='bar')
plt.title('Employee Distribution by department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.show()

sns.histplot(data= ABC_workforce, x='Age', kde=True)
plt.title('Employee Distribution by Age')
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.show()

HR_employee_attrition['Attrition'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Overall Attrition Distribution')
plt.ylabel('')
plt.show()

sns.histplot(data=HR_employee_attrition, x='MonthlyIncome', kde=True)
plt.title('Employee Distribution by Monthly Income')
plt.xlabel('Monthly Income')
plt.ylabel('Number of Employees')
plt.show()
```

### 1. Interpretation

Gender Distribution: The workforce is roughly 60% male and 40% female, showing a moderate gender imbalance rather than an even split.

Department Distribution: Research & Development is the largest department by employee count, followed by Sales, with Human Resources being the smallest.

Age Distribution: The age distribution is roughly bell-shaped and slightly right-skewed, with the largest concentration of employees between ages 27-40, tapering off toward older ages.

Overall Attrition Distribution: About 16% of employees have left the company, while 84% remain, indicating attrition is a real but relatively minority phenomenon in ABC Manufacturing Ltd

Monthly Income Distribution: Monthly income is right-skewed, with most employees earning toward the lower-to-middle end of the range and a smaller group of high earners extending the distribution upward.

```python
#2. Which departments have the highest employee attrition?
attrition_rate = HR_employee_attrition.groupby('Department')['Attrition'].value_counts(normalize=True).unstack()* 100
attrition_rate['Yes'].plot(kind='bar')
plt.title('Attrition Rate by Department')
plt.xlabel('Department')
plt.ylabel('Attrition Rate')
plt.show()
```

## 2.Interpretation
Sales and HR show the highest attrition rates (19-20%), while Research & Development, despite having the highest raw number of departures, actually has the lowest attrition rate (13.5%) due to its much larger size

```python
#3. Does age influence attrition?
sns.boxplot(data=HR_employee_attrition, x='Attrition', y='Age')
plt.title('Age Distribution by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Age')
plt.show()
```

### 3. Interpretation
Employees who left tend to be slightly younger (median age of 32) than those who stayed (median age of 36). The overlap between the two groups is substantial, indicating age alone is a weak predictor of attrition.

```python
#4.Does monthly income affect retention?
sns.boxplot(data=HR_employee_attrition, x='Attrition', y='MonthlyIncome')
plt.title('Monthly Income Distribution by Attrition')
plt.xlabel('Attrition')
plt.ylabel('MonthlyIncome')
plt.show()
```

### 4. Interpretation
Employees who left had a notably lower median income (3,000) than those who stayed (5,000). Higher earners are rarely found among leavers, suggesting income is a meaningful factor in retention.

```python
#5. Does overtime influence attrition?
overtime_rate = HR_employee_attrition.groupby('OverTime')['Attrition'].value_counts(normalize=True).unstack()* 100
overtime_rate.plot(kind='bar')
plt.title('Attrition rate by Overtime status')
plt.xlabel('Overtime')
plt.ylabel('Attrition')
plt.show()
```

### 5. Interpretation
Employees who work overtime have an attrition rate of 30%, three times higher than the 10% rate for those who don't work overtime, making overtime one of the strongest factors linked to attrition.

```python
# 6. Which job roles experience the highest turnover?
Attrition_rate_by_job_role = HR_employee_attrition.groupby('JobRole')['Attrition'].value_counts(normalize=True).unstack()* 100
Attrition_rate_by_job_role['Yes'].plot(kind='bar')
plt.title('Attrition rate by Job Role')
plt.xlabel('Job Role')
plt.ylabel('Attrition')
plt.show()
```

### 6.  Interpretation
Sales Representative shows by far the highest attrition rate (40%), more than double the next highest roles (Laboratory Technician and HR, 23-24%). Seniorroles like Research Director (2.5%) and Manager (5%) show the lowest attrition.

## 7. Key Business Insights

### Five Workforce Observations

1. The company's workforce skews toward early-to-mid career employees, with the largest concentration between ages 27-40, and relatively few employees above 55.
2. Research & Development is the largest department, followed by Sales, with Human Resources being the smallest.
3. The workforce is roughly 60% male and 40% female, indicating a moderate gender imbalance rather than an even split.
4. Monthly income is right-skewed across the workforce, meaning most employees earn toward the lower-to-middle end of the income range, with a smaller number of high earners pulling the distribution upward.
5. Overall attrition across the company stands at approximately 16%, meaning the large majority of employees (84%) remain with the company.

### Five Factors Contributing to Attrition

1. **Overtime** — employees who work overtime have an attrition rate of ~30%, three times higher than those who don't (~10%), suggesting workload is a significant driver of turnover.
2. **Monthly Income** — employees who left had a noticeably lower median income (~₦3,200) than those who stayed (~₦5,200), suggesting compensation plays a real role in retention.
3. **Job Role** — attrition is heavily concentrated in specific roles, particularly Sales Representative (~40%), while senior/specialized roles like Research Director (~2.5%) show very low attrition.
4. **Department** — Sales and HR show the highest attrition rates (~19-20%), while R&D has the lowest (~13.5%), despite having the most total departures in raw numbers.
5. **Age** — a mild pattern exists, with employees who leave skewing slightly younger (median 32) than those who stay (median 36), though this is a weaker signal than the factors above.

### Variables Likely to Matter for Future Predictive Modelling

Based on the patterns above, **OverTime**, **MonthlyIncome**, and **JobRole** appear to be the strongest candidate predictors for a future attrition model, each showing a clear, sizeable gap between employees who left and those who stayed. **Department** shows a moderate relationship, largely a broader reflection of the JobRole signal. **Age** shows only a weak relationship and would likely need to be combined with other variables to add real predictive value.
